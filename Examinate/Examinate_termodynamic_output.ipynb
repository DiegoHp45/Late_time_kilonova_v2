```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
main_out=pd.read_csv('Example_NSM_dyn_ejecta_rosswog_500dias_sin_theoretical_beta_alpha_out/mainout.dat',
                     delim_whitespace=True,
                     skiprows=3,
                     header=None,
                     names=['iteration','time[s]','T[GK]','rho[g/cm^3]','Ye','R[km]','Y_n','Y_p','Y_alpha','Y_gamma','Y_heavies','<Z>','<A>','entropy[kb/baryon]'])
```

```python
plt.figure(figsize=(5,4))
plt.scatter(main_out['time[s]']/(24*60*60),main_out['T[GK]'],s=1)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('time [days]')
plt.ylabel('T [GK]')
```

```python
plt.figure(figsize=(5,4))
plt.scatter(main_out['time[s]']/(24*60*60),main_out['rho[g/cm^3]'],s=1)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('time [days]')
plt.ylabel('density[g/cm^3]')
```

```python
plt.figure(figsize=(5,4))
plt.scatter(main_out['time[s]']/(24*60*60),main_out['R[km]'],s=1)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('time [days]')
plt.ylabel('R [km]')
```
