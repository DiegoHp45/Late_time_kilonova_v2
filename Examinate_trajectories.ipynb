```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
Thermodynamic_trayectory_file='Nuclear_data/Thermodynamic_trajectories/nsm_representative.dat'
trayectory_data = pd.read_csv(Thermodynamic_trayectory_file,header=0,delim_whitespace=True,usecols=['time[s],','T9[GK],','density[g/cm^3],','R[km],','Ye'])
```

```python
plt.figure(figsize=(5,4))
plt.ylabel('R[km]')
plt.xlabel('time[s]')
plt.scatter(trayectory_data['time[s],'],trayectory_data['R[km],'],s=1)
```

```python
plt.figure(figsize=(5,4))
plt.scatter(trayectory_data['time[s],'],trayectory_data['density[g/cm^3],'],s=1)
plt.yscale('log')
plt.xscale('log')
plt.ylabel(r'density[$g cm^{-3}$]')
plt.xlabel('time[s]')
```

```python
plt.figure(figsize=(5,4))
plt.scatter(trayectory_data['time[s],'],trayectory_data['T9[GK],'],s=1)
plt.yscale('log')
plt.xscale('log')
plt.ylabel(r'T[$GK$]')
plt.xlabel('time[s]')
```
