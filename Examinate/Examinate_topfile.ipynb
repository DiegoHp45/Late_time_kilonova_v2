```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
toplist_producers = pd.read_fwf(
    'Example_NSM_dyn_ejecta_rosswog_500dias_sin_theoretical_beta_alpha_out/toplist.dat',
    widths=[12,16,37,33,33,33,33,33,33,33,33,37,15,15,15,15,15,15,15,15,15,18],
    skiprows=1,
    header=0,
)
```

```python
toplist_consumers = pd.read_fwf(
    'Example_NSM_dyn_ejecta_rosswog_500dias_sin_theoretical_beta_alpha_out/toplist.dat',
    widths=[519,33,33,33,33,33,33,33,33,33,33,15,15,15,15,15,15,15,15,15,15],
    skiprows=1,
    header=0,
)
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(toplist_producers['#    Time[s]'], toplist_producers['Etot[erg/g/s]'],s=0.5)
plt.xlabel('Time (s)')
plt.ylabel('Energy Generation Rate (erg/g/s)')
plt.scatter(toplist_producers['#    Time[s]'], toplist_producers['Ereac_1'],s=5)
plt.scatter(toplist_producers['#    Time[s]'], toplist_consumers['Ereac_1']*-1,s=0.1)
plt.yscale('log')
plt.xscale('log')
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(toplist_producers['#    Time[s]']/(24*60*60), toplist_producers['Etot[erg/g/s]'],s=0.5)
plt.xlabel('Time (s)')
plt.ylabel('Energy Generation Rate (erg/g/s)')
plt.scatter(toplist_producers['#    Time[s]']/(24*60*60), toplist_producers['Ereac_1'],s=0.5)
plt.yscale('log')
plt.xscale('log')
plt.xlim(1e-5,500)
plt.ylim(1e4,1e20)
```
