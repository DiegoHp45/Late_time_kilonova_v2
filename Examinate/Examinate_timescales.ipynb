```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
timescales_out=pd.read_csv('Example_NSM_dyn_ejecta_rosswog_500dias_sin_theoretical_beta_alpha_out/timescales.dat',
                            delim_whitespace=True,
                            skiprows=1,
                            header=None,
                            names=['time[s]', 'Temperature [GK]', 'tau_ga [s]', 'tau_ag [s]', 'tau_ng [s]', 'tau_gn [s]', 'tau_pg [s]', 'tau_gp [s]', 'tau_np [s]', 'tau_pn [s]', 'tau_an [s]', 'tau_na [s]', 'tau_ap [s]', 'tau_pa [s]', 'tau_beta [s]', 'tau_alpha [s]', 'tau_nfiss [s]', 'tau_sfiss [s]', 'tau_bfiss [s]'])
```

```python
plt.figure(figsize=(10, 6))
plt.loglog(timescales_out['time[s]'], timescales_out['tau_alpha [s]'], label='tau_alpha')
plt.loglog(timescales_out['time[s]'], timescales_out['tau_beta [s]'], label='tau_beta')
plt.loglog(timescales_out['time[s]'], timescales_out['tau_sfiss [s]'], label='tau_sfiss')
```
