```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
%matplotlib widget
```

```python
Energy_out=pd.read_csv('Example_NSM_dyn_ejecta_rosswog_500dias_sin_theoretical_beta_alpha_out/generated_energy.dat',
                       delim_whitespace=True,
                       skiprows=2,
                       header=None,
                       names=['time[s]', 'Engen(Total)', 'S_src', 'Engen(weak)', 'Engen(n,g)', 'Engen(p,g)', 'Engen(a,g)', 'Engen(n,p)', 'Engen(n,a)', 'Engen(p,a)', 'Engen(fiss)'])
```

```python
plt.figure(figsize=(5,4))
plt.scatter(Energy_out['time[s]']/86400, Energy_out['Engen(Total)'], s=1, label='Total')
plt.scatter(Energy_out['time[s]']/86400, Energy_out['Engen(weak)'], s=1, label='Weak')
plt.scatter(Energy_out['time[s]']/86400, Energy_out['Engen(fiss)'], s=1, label='fission')


plt.ylabel('Time[Days]')
plt.ylabel('Energy generation rate [erg/g/s]')
plt.yscale('log')
plt.xscale('log')
plt.xlim(0.00001, 500)
plt.ylim(1e4, 1e20)
plt.legend()
```
